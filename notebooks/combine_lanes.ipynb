```python
import glob
import os
import pandas as pd
```

```python
lane1_fastq_files = sorted(glob.glob('/data/shared/2022_10_3_WildBeer_NovaSeqMG/22419Fie_N22166/*L001*.fastq.gz'))
lane2_fastq_files = sorted(glob.glob('/data/shared/2022_10_3_WildBeer_NovaSeqMG/22419Fie_N22166/*L002*.fastq.gz'))
```

```python
intermediate_path = '/data/shared/2022_10_3_WildBeer_NovaSeqMG/intermediate/'
for l1_file, l2_file in list(zip(lane1_fastq_files, lane2_fastq_files))[1:]:
    l1_basename = l1_file.split('/')[-1]
    l1_fname = l1_basename.split('_L00')[0]
    
    l2_basename = l2_file.split('/')[-1]
    l2_fname = l2_basename.split('_L00')[0]
    
    print(l1_basename, l2_basename)

    
    fwd_or_reverse = l1_basename.split('_R')[-1].split('_')[0]
    parsed_fname = '_'.join(l1_fname.split('_')[1:]) + f'_R{fwd_or_reverse}'
    
    print(parsed_fname)
    
    if l1_fname != l2_fname:
        print('uh oh!')
        break
        
    print('copying')
    os.system(f'cp {l1_file} {intermediate_path}')
    os.system(f'cp {l2_file} {intermediate_path}')
    print('unzipping')
    os.system(f'gunzip {intermediate_path}*{l1_fname}*')
    
    print('concatenating')
    os.system(f'cat {intermediate_path}*{l1_fname}* > {intermediate_path}{parsed_fname}.fastq')
    
    print('checking file lengths')
    result = os.system(f'wc -l {intermediate_path}*')
    print(result)
    print('rezipping')
    os.system(f'gzip {intermediate_path}{parsed_fname}.fastq')
    print('moving')
    os.system(f'mv {intermediate_path}{parsed_fname}.fastq.gz /data/shared/2022_10_3_WildBeer_NovaSeqMG/combined_lanes/')
    os.system(f'rm {intermediate_path}*')
        
        
```
