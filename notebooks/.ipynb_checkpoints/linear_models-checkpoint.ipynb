```python
# add additional experiment columns back
chem_wide_scaled['timepoint'] = chem_wide_scaled.index.to_series().apply(lambda x: '0' if 'C' in x else x.split('-')[-1])
chem_wide_scaled['hops'] = chem_wide_scaled.index.to_series().apply(lambda x: 1 if x[0] == 'H' else 0)
exposure_map = {'C':0, 'I':1, 'O':2}
chem_wide_scaled['exposure'] = chem_wide_scaled.index.to_series().apply(lambda x: exposure_map[x[1]])
chem_wide_scaled['treatment'] = chem_wide_scaled.index.to_series().apply(lambda x: x[:2])
chem_wide_scaled['replicate'] = chem_wide_scaled.index.to_series().apply(lambda x: x.split('-')[0][-1])
chem_wide_scaled.head(2)
```

## Load and process RPKM data

```python
# load pre-computed MAG coverage table
table = pd.read_csv('/data/mhoffert/fiererlab/beer/data/20230605_MAG_coverage_table.tsv', sep='\t', index_col=0)
table = table.drop(['CommStd_S36',
           'ExtB1_S43',
           'NTC_S54'], axis=0)

# set control RPKMs to 1 in controls for purposes of regression
table[table.index.str.contains('C')] = 0
```

```python
# filter prevalence of species at 1 RPKM
filtered_taxa =  table.columns[(table > 1).sum() > 10]

clr_table = pd.DataFrame(clr(table.loc[:, filtered_taxa] + 1e-1),
                         index=table.index,
                         columns=filtered_taxa)
```

```python
# clean up MG names
clr_table = clr_table.rename(index=dict((i, i.split('_')[0]) for i in clr_table.index))
```

```python
clr_table.head()
```

# Perform regression

```python
# reg_df = pd.merge(clr_table, chem_wide_scaled, left_index=True, right_index=True)
# reg_df.head()
```

```python
def apply_fdr(results_df, alpha=0.05):
    valid = results_df["p_value"].notna()
    results_df["q_value"] = np.nan
    results_df.loc[valid, "q_value"] = multipletests(
        results_df.loc[valid, "p_value"], alpha=alpha, method="fdr_bh"
    )[1]
    results_df["significant"] = results_df["q_value"] < alpha
    return results_df.sort_values("q_value")
```

```python
# taxon = 'g__Enterobacter'
results = []
for taxon in clr_table.columns:
    for chem_var in chem_wide_scaled.iloc[:, 0:7].columns:
        # print(f'Variable: {chem_var}')
        # print(f"Taxon: {taxon}")
        design = chem_wide_scaled[["hops", "exposure", "timepoint", 'replicate']].copy()
        for col in design.columns:
            if col in ['exposure', 'hops', 'replicate']:
                design[col] = design[col].astype('category')
            else:
                design[col] = design[col].astype(float)
        design['replicate'] = design.index.to_series().apply(lambda x: x.split('-')[0])
        # design["exposure"] = design["exposure"].astype("category")
        # design["hops"] = design["hops"].astype("category")

        y = chem_wide_scaled.loc[:, chem_var]
        df = design.copy()
        df["y"] = y
        df["taxon_abund"] = clr_table.loc[design.index, taxon]
        # df['int'] = 1
        df = df[~df.exposure.eq(0)]
        df = df.dropna()

        formula = "y ~ taxon_abund + hops * exposure + timepoint"

        # model = MixedLM.from_formula(
        #     formula, 
        #     groups='replicate', 
        #     data=df
        # )
        
        model = smf.ols(formula, data=df)
        
        fit = model.fit(reml=True, method="pinv")
        
        results.append({
            "chem_var": chem_var,
            "taxon": taxon,
            "coef": fit.params.get("taxon_abund", np.nan),
            "se": fit.bse.get("taxon_abund", np.nan),
            "p_value": fit.pvalues.get("taxon_abund", np.nan),
            # "converged": fit.converged,
            'log_likelihood': fit.llf
        })
    
results_df = pd.DataFrame(results)
```

## Species vs experimental design models

```python
print('\n'.join(list(clr_table.columns)))
tax_map = {'S_cerevisiae':'Y',
            'g__Enterobacter':'E',
            's__Lacticaseibacillus paracasei':'L',
            's__Lentilactobacillus parabuchneri':'L',
            's__Pantoea agglomerans':'E',
            's__Pseudescherichia vulneris':'E',
            's__Schleiferilactobacillus harbinensis':'L'}
```

```python
tax_clr_table = clr_table.groupby(tax_map, axis=1).sum()
```

```python
# fit 1 model per clade: E, L, Y
results = []
fitted_models = {}

design = chem_wide_scaled[["hops", "exposure", "timepoint", 'replicate']].copy()
design['exposure'] = design['exposure'] - 1
design = design[~design['exposure'].eq(-1)]

for col in design.columns:
    if col in ['exposure', 'hops']:
        design[col] = design[col].astype('category')
    else:
        design[col] = design[col].astype(float)

for taxon in tax_clr_table.columns:

    df = design.copy()
    df["taxon_abund"] = tax_clr_table.loc[design.index, taxon]
    df = df[~df.exposure.eq(-1)]
    df = df.dropna()

    formula = "taxon_abund ~ hops * exposure + timepoint"

    # model = MixedLM.from_formula(
    #     formula, groups='replicate', data=df
    # )
    # fit = model.fit(reml=True, method="powell")
    
    model = smf.ols(formula = formula,
                    data=df
                   )
    
    fit = model.fit(reml=True, method='pinv')

    fitted_models[taxon] = fit
    
    fit_df = pd.concat([fit.params, fit.bse, fit.pvalues], 
          keys=['coef', 'se', 'p_value'], axis=1).assign(taxon=taxon, 
                                                #converged=fit.converged,
                                                ll=fit.llf)

    results.append(fit_df)
    
    
sum_results_df = pd.concat(results)
```

## Residuals (effect of hops and exposure in decreasing variation)

```python
fitted_models.keys()
```

```python
# make design df once
design = chem_wide_scaled[["hops", "exposure", "timepoint"]].copy()
design['exposure'] = design['exposure'] - 1
design = design[~design.exposure.eq(-1)]
for col in design.columns:
    if col in ['exposure', 'hops']:
        design[col] = design[col].astype('category')
    else:
        design[col] = design[col].astype(float)

# objs to save results
resid_fitted_models = {}    
resid_results = []

# run residual model for each taxon
for taxon in clr_column_order:
    
    df = design.copy()
    df = df.dropna()
    df = df[~df['exposure'].eq(-1)]
    # add residual column
    resid = fitted_models[taxon].resid.abs().reindex(df.index)
    df['resid'] = resid
    df['jar'] = df.index.to_series().apply(lambda x: x[:3])
    df['treatment'] = df.index.to_series().apply(lambda x: x[:2])

    resid_fitted_models[taxon] = {}
    for timepoint in [1, 3, 'both']:
        
        if timepoint != 'both':
            
            resid_model = smf.ols(formula = 'resid ~ hops * exposure', 
                              data=df[df['timepoint'].eq(timepoint)])
        else:
            mean_df = df.groupby('jar').agg({'hops':'first', 'exposure':'first', 'resid':'mean'})
            
            resid_model = smf.ols(formula = 'resid ~ hops * exposure', 
                              data=mean_df)
        # fit
        resid_fit = resid_model.fit()

        resid_fitted_models[taxon][str(timepoint)] = resid_fit

        fit_df = pd.concat([resid_fit.params, resid_fit.bse, resid_fit.pvalues], 
              keys=['coef', 'se', 'p_value'], axis=1).assign(taxon=taxon,
                                                             timepoint=timepoint,
                                                    # converged=fit.converged,
                                                    ll=resid_fit.llf)

        resid_results.append(fit_df)
```
