```python
def edit_distance_myers(seq1: str, seq2: str) -> int:
    """
    Myers' bit-vector algorithm for edit distance.
    O(n*m/w) time where w=64 (word size). Much faster than DP for long sequences.
    """
    if seq1 == seq2:
        return 0
    if len(seq1) == 0:
        return len(seq2)
    if len(seq2) == 0:
        return len(seq1)

    # Ensure pattern (seq1) is the shorter for peq masks, text (seq2) is longer
    if len(seq1) > len(seq2):
        seq1, seq2 = seq2, seq1

    n = len(seq1)
    m = len(seq2)

    # Build character equality bitmasks (Peq) for A, C, G, T, N
    peq = {c: 0 for c in "ACGTN"}
    for i, ch in enumerate(seq1):
        peq.setdefault(ch, 0)
        peq[ch] |= (1 << i)

    mask = 1 << (n - 1)
    vp = (1 << n) - 1  # all 1s
    vn = 0
    score = n

    for ch in seq2:
        eq = peq.get(ch, 0)  # unknown chars match nothing
        xv = eq | vn
        xh = (((eq & vp) + vp) ^ vp) | eq

        ph = vn | ~(xh | vp)
        mh = vp & xh

        # Keep within n bits
        full_mask = (1 << n) - 1
        ph &= full_mask
        mh &= full_mask

        if ph & mask:
            score += 1
        elif mh & mask:
            score -= 1

        ph_shift = (ph << 1) | 1
        mh_shift = mh << 1

        vp = (mh_shift) | ~(xv | ph_shift)
        vn = ph_shift & xv

        vp &= full_mask
        vn &= full_mask

    return score
    
```

```python
def count_segments(data):

    # Count of True segments
    true_segments = np.sum([1 for key, group in groupby(data) if key])
    return true_segments

def palindrome_score(seq: str) -> float:
    """
    Score a single window for palindrome-ness.
    1.0 = perfect palindrome (seq == rc(seq))
    0.0 = maximally different from its own reverse complement
    """
    rc = str(Seq(seq).reverse_complement())
    ed = edit_distance_myers(seq, rc)
    return 1.0 - (ed / len(seq))

def scan_palindrome_scores(dna_seq, min_ws, max_ws):
    """
    For each window size in [min_ws, max_ws), score every position for
    self-palindrome-ness. Returns the max score per position across window
    sizes (best-fitting window), plus which window size achieved it.
    """
    seq_str = str(dna_seq)
    n = len(seq_str)
    all_scores = np.full((max_ws - min_ws, n), np.nan)

    for k, ws in enumerate(range(min_ws, max_ws)):
        for i in range(n - ws + 1):
            all_scores[k, i] = palindrome_score(seq_str[i:i+ws])

    best_scores = np.nanmax(np.nan_to_num(all_scores, 0), axis=0)
    best_ws_idx = np.nanargmax(np.nan_to_num(all_scores, nan=-1), axis=0)
    best_ws = np.where(np.isnan(best_scores), np.nan, best_ws_idx + min_ws)

    return best_scores, best_ws

def palindrome_score_zscore(seq: str, n_null: int = 200) -> float:
    """
    Z-score of observed palindrome edit distance vs. random sequences
    of the same length. More negative = more unexpectedly palindromic.
    Converted to a 0-1 "percentile-like" score via a sigmoid squash.
    """
    ws = len(seq)
    rc = str(Seq(seq).reverse_complement())
    obs_ed = edit_distance_myers(seq, rc)

    null_eds = np.array([
        edit_distance_myers(random_dna(ws), str(Seq(random_dna(ws)).reverse_complement()))
        for _ in range(n_null)
    ])
    mu, sigma = null_eds.mean(), null_eds.std()
    z = (mu - obs_ed) / sigma if sigma > 0 else 0.0

    # squash z to (0,1); z=0 -> 0.5, more negative ed (more palindromic) -> closer to 1
    return 1 / (1 + np.exp(-z))
    
```

```python
# code for making test dataset
def random_dna(n: int) -> str:
    '''
    make a random DNA sequence
    '''
    return ''.join(np.random.choice(list('ACGT'), size=n))

def make_pseudopalindrome(half_len: int, mismatch_rate: float = 0.0, spacer: int = 0) -> str:
    """
    make a palindromic seq: half + spacer + rc(half), 
    mismatch_rate determines mismatches in introduced into the second half.
    """
    half = random_dna(half_len)
    rc_half = list(str(Seq(half).reverse_complement()))
    
    # number of mismatches to add
    n_mismatch = int(round(half_len * mismatch_rate))
    # if it isn't zero, catch bug
    if n_mismatch > 0:
        positions = np.random.choice(half_len, size=n_mismatch, replace=False)
        # make random changes
        for p in positions:
            alt_bases = [b for b in 'ACGT' if b != rc_half[p]]
            rc_half[p] = np.random.choice(alt_bases)
            
    # make a random insert
    spacer_seq = random_dna(spacer)
    return half + spacer_seq + ''.join(rc_half)

def generate_palindrome_dataset(
    n_per_condition: int = 20,
    background_len: int = 200,
    half_lens=(8, 9, 10),          # ~16-20bp full site, MarR-like
    mismatch_rates=np.linspace(0, 1, 20),
    spacers=(0, 1, 2, 3),
    include_negatives: bool = True,
    seed: int = None,
):
    """
    make a test dataset: written with Claude
    
    Returns a list of dicts:
      seq: SeqRecord (background with embedded IR)
      start, length: 0-indexed position and total length of the embedded motif
      half_len, mismatch_rate, spacer: generation params
      label: 'palindrome' or 'negative'
    """
    if seed is not None:
        np.random.seed(seed)

    records = []
    idx = 0
    for half_len in half_lens:
        for mismatch_rate in mismatch_rates:
            for spacer in spacers:
                for _ in range(n_per_condition):
                    motif = make_pseudopalindrome(half_len, mismatch_rate, spacer)
                    motif_len = len(motif)

                    bg_len = background_len - motif_len
                    insert_pos = np.random.randint(0, bg_len)
                    bg = random_dna(bg_len)
                    full_seq = bg[:insert_pos] + motif + bg[insert_pos:]

                    records.append({
                        'seq': SeqRecord(Seq(full_seq), id=f'pal_{idx}'),
                        'start': insert_pos,
                        'length': motif_len,
                        'half_len': half_len,
                        'mismatch_rate': mismatch_rate,
                        'spacer': spacer,
                        'label': 'palindrome',
                    })
                    idx += 1

    if include_negatives:
        for _ in range(n_per_condition * len(half_lens)):
            full_seq = random_dna(background_len)
            records.append({
                'seq': SeqRecord(Seq(full_seq), id=f'neg_{idx}'),
                'start': None,
                'length': None,
                'half_len': None,
                'mismatch_rate': None,
                'spacer': None,
                'label': 'negative',
            })
            idx += 1

    return records
```

```python
def print_palindrome_scan(dna_seq, min_ws=6, max_ws=12, min_distance=5,
                           n_null=200, alpha=0.05, line_width=80, seed=None):
    '''
    Run scan_palindrome_scores and print the sequence with an ASCII track
    underneath marking detected palindrome positions.

    Peaks are called using a length-specific null threshold (random sequences
    of the matching window size) rather than a flat score cutoff, since short
    windows hit high raw scores by chance much more easily than long ones.

    Symbols:
      '=' : part of a detected palindrome window
      '^' : peak center
      '.' : below threshold / not part of a call
    '''
    if seed is not None:
        np.random.seed(seed)

    seq_str = str(dna_seq)
    n = len(seq_str)

    best_scores, best_ws = scan_palindrome_scores(dna_seq, min_ws, max_ws)

    # build a length-specific score threshold from a null (random-sequence) distribution,
    # since a 6bp window and a 11bp window do NOT hit a given score by chance equally often
    null_thresh = {}
    for ws in range(min_ws, max_ws):
        null_scores = np.array([palindrome_score(random_dna(ws)) for _ in range(n_null)])
        null_thresh[ws] = np.quantile(null_scores, 1 - alpha)

    # per-position threshold, based on whichever window size was "best" there
    pos_thresh = np.array([null_thresh.get(int(ws), np.inf) if not np.isnan(ws) else np.inf
                            for ws in best_ws])

    finite_scores = np.nan_to_num(best_scores, nan=0.0)
    passes = finite_scores >= pos_thresh

    # peak call only among positions that clear their length-specific null threshold
    candidate_idx = np.nonzero(passes)[0]
    if len(candidate_idx) == 0:
        peak_idx = np.array([], dtype=int)
    else:
        # find_peaks needs a continuous array; mask failing positions to -inf so they're never peaks
        masked_scores = np.where(passes, finite_scores, -np.inf)
        peak_idx, _ = find_peaks(masked_scores, distance=min_distance)
        # find_peaks won't call a peak at the very ends; also keep endpoints if they clear threshold
        # and are local-ish maxima among candidates (optional, skipped here for simplicity)

    # build indicator track, respecting existing marks so overlaps don't silently overwrite
    track = np.array(['.'] * n, dtype='<U1')
    for i in peak_idx:
        ws = int(best_ws[i])
        end_ = min(n, i + ws)
        for pos in range(i, end_):
            if track[pos] == '.':
                track[pos] = '='
    for i in peak_idx:
        track[i] = '^'

    for chunk_start in range(0, n, line_width):
        chunk_end = min(n, chunk_start + line_width)
        seq_chunk = seq_str[chunk_start:chunk_end]
        track_chunk = ''.join(track[chunk_start:chunk_end])
        print(f'{chunk_start:>6} {seq_chunk}')
        print(f'{"":>6} {track_chunk}')
        print()

    peak_scores = finite_scores[peak_idx]
    peak_ws_out = best_ws[peak_idx].astype(int)
    return peak_idx, peak_scores, peak_ws_out
```

```python
def write_pali_promos(df, file):
    lines = []
    for i, (index, row) in enumerate(df.iterrows()):
        lines.append(f'>{row["seq_id"]}')
        lines.append(f'{row["sequence"]}')
    with open(file, 'w') as handle:
        handle.write('\n'.join(lines))
```

```python
def get_nearby_orfs(genome_fasta, contig_order, loc_info):
    '''
    get orfs from concatenated 
    '''
    
    with open(genome_fasta, 'r') as handle:
        records = sorted([r for r in SeqIO.parse(handle, 'fasta')], key=lambda r: contig_order.index(r.id))
    curr_pos = 0
    i = 0
    while loc_info.loc['start'] > curr_pos:
        curr_pos = curr_pos + len(records[i].seq)
        i += 1
    
    chr_start = len(records[i-1].seq) - int(curr_pos - loc_info.loc['start'])
    loc_len = int(loc_info.loc['stop'] - loc_info.loc['start'])
    print(f'In: contig {records[i-1].id}')
    print(f'Chr start: {chr_start}')
    seq = records[i-1].seq[chr_start-1:chr_start+loc_len+1]
    
    strand = 1
    if loc_info.loc['strand'] == '+':
        found_seq = seq
        strand = 1
    else:
        found_seq = seq.reverse_complement()
        strand = -1
        
    print(strand)
    return loc_info.loc['id'], records[i-1].id, chr_start, found_seq, strand
    
```

```python
from collections import Counter
```

```python
def summarize_annotations(annotations, top_n=5):
    """
    annotations: list of lists, e.g. [['Formamidopyrimidine-DNA glycosylase'], ['DNA glycosylase/AP lyase', 'H2TH DNA-binding'], ...]
    Returns a concise summary string.
    """
    # Flatten, clean, drop placeholders
    flat = [a.strip() for group in annotations for a in group if a.strip() and a.strip() != '-']
    
    if not flat:
        return "Uncharacterized protein"
    
    counts = Counter(flat)
    
    # Prefer the annotation that looks like a full protein/family name
    # (heuristic: no "domain"/"finger"/"site" as a *second* fragment, i.e. it's a first element)
    primary_candidates = [group[0].strip() for group in annotations if group[0].strip() != '-']
    primary_counts = Counter(primary_candidates)
    primary_name = primary_counts.most_common(1)[0][0]
    
    # Collect distinct domain descriptors (excluding the primary name itself)
    domain_terms = [term for term in flat if term != primary_name]
    domain_counts = Counter(domain_terms)
    top_domains = [term for term, _ in domain_counts.most_common(top_n)]
    
    summary = f"{primary_name}"
    if top_domains:
        summary += f" ({', '.join(top_domains)})"
    return summary
```

```python
from pygenomeviz import GenomeViz
from pygenomeviz.parser import Gff
from pygenomeviz.utils import load_example_gff_file

gff_file = load_example_gff_file("escherichia_coli.gff.gz")
gff = Gff(gff_file)

gv = GenomeViz()
gv.set_scale_bar(ymargin=0.5)

target_ranges = ((215000, 230000), (300000, 320000))
track = gv.add_feature_track(name=gff.name, segments=target_ranges)
track.set_segment_sep()

for segment in track.segments:
    segment.add_sublabel()
    features = gff.extract_features(feature_type=None, target_range=segment.range)
    for feature in features:
        if feature.type == "CDS":
            segment.add_features(feature, label_type="gene", fc="skyblue", lw=1.0, annotation=True, text_kws=dict(bbox=dict(boxstyle="round", fc="skyblue")))
        elif feature.type == "rRNA":
            segment.add_features(feature, label_type="product", fc="lime", lw=1.0, hatch="//", annotation=True, text_kws=dict(bbox=dict(boxstyle="round", fc="lime")))

gv.savefig("gff_features.png")
```
