```python
def stacked_bar(tidy_data, x, y, hue_col, order, palette_map, ax):
    hues = tidy_data[hue_col].unique()
    piv = tidy_data.pivot_table(index=hue_col, columns=x, values=y)
    cumsum = piv.reindex(index=order).cumsum().stack().reset_index().rename(columns={0:y})

    for hue_item, z in zip(order, range(len(order), 0, -1)):
        sns.barplot(data=cumsum[cumsum.species.eq(hue_item)], x=x, y=y, color=palette_map[hue_item], zorder=z, ax=ax)
```

```python
# color_map = dict(zip(order, ['green',  'skyblue', 'blue', 'darkblue', 'red', 'gray', 'lightcoral', 'dodgerblue']))
species_color_map = {'s__Gluconobacter_A japonicus': 'green',
 's__Schleiferilactobacillus harbinensis': 'skyblue',
 's__Pantoea agglomerans':'firebrick',
 's__Levilactobacillus brevis': 'darkblue',
 's__Lentilactobacillus parabuchneri': 'dodgerblue',
 'g__Enterobacter': 'lightcoral', 
 'S_cerevisiae': 'gray',
 's__Pseudescherichia vulneris': 'red',
 's__Lacticaseibacillus paracasei':'blue'}
```

```python
# do it manually
fig = plt.figure(figsize=(17, 9))
i = 1
axes = []
for timepoint, treatment in itertools.product(['1 week', '3 week'], ['HI', 'HO', 'UI', 'UO']):
    print(treatment, timepoint)
    subset = strip_plotdf[strip_plotdf['timepoint'].eq(timepoint) & strip_plotdf['treatment'].eq(treatment)]


    ax = fig.add_subplot(2,4,i)
    #     axes.append(ax)
    # else:
    #     ax = fig.add_subplot(2,4, i, sharey=axes[i-2])
    #     axes.append(ax)
    
    
    subset_plotdf = subset.pivot_table(columns='replicate', index='species', values='rpkm').T
    stacked_bar(subset, 'replicate', 'rpkm', 'species', species_color_map, ax)
    
    ax.set_ylabel('RPKM', fontsize=16)
    ax.set_xlabel('Replicate', fontsize=16)
    
    sns.despine()
    
    plt.tick_params(labelsize=14)
    if i < 5:
        plt.tick_params(bottom=False, labelbottom=False)
        ax.set_xlabel('')
    
    if i % 4 != 1:
        # plt.tick_params(left=False, labelleft=False)
        ax.set_ylabel('')
        
    upper =  int(np.floor(ax.get_ylim()[1] / 100))
    for j in np.linspace(100, upper * 100, upper):
        ax.axhline(j, zorder=0, color='k', alpha=1, linewidth=0.5)
        
    i += 1

annot_kws = {'xycoords':'figure fraction', 'fontsize':20, 'ha':'center'}
ax.annotate('Hopped', (0.3, 0.95), **annot_kws)
ax.annotate('Unhopped', (0.7, 0.95), **annot_kws)

ax.annotate('Indoor', (0.2, 0.9), **annot_kws)
ax.annotate('Outdoor', (0.4, 0.9), **annot_kws)

ax.annotate('Indoor', (0.6, 0.9), **annot_kws)
ax.annotate('Outdoor', (0.8, 0.9), **annot_kws)

annot_kws = {'xycoords':'figure fraction', 'fontsize':20}

ax.annotate('1 week', (0, 0.6), **annot_kws)
ax.annotate('3 week', (0, 0.3), **annot_kws)


fig.subplots_adjust(hspace=0.1, wspace=0.25)
patches = [mpatches.Patch(color=value, label=key) for key, value in species_color_map.items()]
lgd = plt.legend(handles=patches, loc='center left', bbox_to_anchor=(1, 1))
```

### Heatmap

```python
species_order = ['S_cerevisiae',
 'g__Enterobacter',
 's__Pantoea agglomerans',
 's__Pseudescherichia vulneris',
 's__Lacticaseibacillus paracasei',
 's__Schleiferilactobacillus harbinensis',
 's__Lentilactobacillus parabuchneri',
 's__Levilactobacillus brevis',
 's__Gluconobacter_A japonicus']
```

```python
strip_plotdf['rpkm'].apply(lambda x: np.log10(x) if x > 0 else np.log10(0.01)).min()
```

```python
fig = plt.figure(figsize=(20, 10))
i = 1
for s_i, s in enumerate(species_order):
    for t_i, t in enumerate(['HI', 'HO', 'UI', 'UO']):
        display(i)
        clear_output(wait=True)
        ax = fig.add_subplot(9, 4, i)
        
        subset = strip_plotdf[strip_plotdf.species.eq(s) & 
                              strip_plotdf['timepoint'].isin(['1 week', '3 week']) &
                              strip_plotdf['treatment'].eq(t)].copy(deep=True)
        subset['rpkm'] = subset['rpkm'].apply(lambda x: np.log10(x) if x > 0 else np.log10(1e-5))
        heatmap_data = subset.pivot_table(index='timepoint', columns='replicate', values='rpkm')
        
        if i % 4 == 0:
            cbar=True
        else:
            cbar=False
        
        sns.heatmap(heatmap_data, ax=ax, vmin=0, vmax=2, cbar=cbar, cmap='gnuplot')
        
        if i < 5:
            ax.set_title(f'{t}', fontsize=16)
        
        if i % 4 == 1:
            ylabel = s.replace("s__", "").replace(" ", "\n")
            ax.set_ylabel(f'{ylabel}', fontsize=16, rotation=0, ha='right', va='center')
        else:
            ax.set_ylabel('')
            plt.tick_params(left=False, labelleft=False)
        
        if i < 33:
            ax.set_xlabel('')
            plt.tick_params(bottom=False, labelbottom=False)
            

    
    
    
        i += 1
    
    
```
