```python
import pandas as pd
import glob
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
ssu_tables = glob.glob('/data/ftp/2022_10_3_WildBeer_NovaSeqMG/PhyloFlash/*/*extractedSSUclassifications.csv')

basename = lambda x, ext: x.split('/')[-1].split(ext)[0]
dataframe = pd.concat([pd.read_csv(n).assign(sample=basename(n, '.phyloFlash').split('_')[0]) for n in ssu_tables], axis=0).reset_index(drop=True)
```

```python
dataframe.to_csv('/data/ftp/2022_10_3_WildBeer_NovaSeqMG/PhyloFlash/2022_10_20_WildBeer.phyloFlash.extractedSSUclassifications.csv')
```

```python
dataframe['sample'].unique()
```

```python
lab_aab = ['Gluconobacter oxydans',
 'Lactobacillus brevis ATCC 367',
 'Lactobacillus fermentum',
 'Lactobacillus harbinensis',
 'Lactobacillus harbinensis DSM 16991',
 'Lactobacillus parabuchneri',
 'Lactobacillus paracasei']
lab_aab_dbHits = dataframe[dataframe.taxonomy.apply(lambda x: x.split(';')[-1]).isin(lab_aab)]['dbHit'].unique()
samples = [s for s in dataframe['sample'].unique() if not s in ['CommStd', 'NTC', 'ExtB1']]
```

```python
len(samples)
```

```python
dataframe_wide = dataframe.set_index([ 'dbHit', 'sample'])['read_cov'].unstack().fillna(0)
```

```python
total_ab_abundance = dataframe_wide.divide(dataframe_wide.sum(0)).reindex(index=lab_aab_dbHits, columns=samples).sum()
```

```python
md = pd.read_csv('/data/mhoffert/fiererlab/beer/project_data_and_tutorials/data/2022_10_12_WildBeer_metadata.tsv', sep='\t', index_col=0)
```

```python
md[['Sample ID', 'timepoint', 'pH', 'Location', 'Hops']] #.to_csv('2022_10_21_WildBeer_md_for_R.tsv', sep='\t')
```

```python
len(total_ab_abundance)
```

```python
data_for_r = pd.merge(md[['Sample ID', 'timepoint', 'pH', 'Location', 'Hops']],
         total_ab_abundance.rename('AB_abundance'), right_index=True, left_on='Sample ID')
```

```python
data_for_r['timepoint'] = data_for_r['timepoint'].map({'1 week':1,'initial':0, '3 week':3})
data_for_r['Outside'] = data_for_r['Location'].map({'C':0,'I':0, 'O':1})
data_for_r['Hops'] = data_for_r['Hops'].map({'H':1,'U':0})
```

```python
data_for_r[['Sample ID', 'timepoint', 'pH', 'Outside', 'Hops', 'AB_abundance']].to_csv('2022_10_21_WildBeer_md_for_R.tsv', sep='\t')
```
